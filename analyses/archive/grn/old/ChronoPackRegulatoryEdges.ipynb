# ChronoPack GRN Edges

Builds clean ChronoPack TF-target edge predictions from recalibrated target-gene scores and OmniPath TF priority scores. Benchmark-specific TF-universe filtering is intentionally not part of this active inference notebook.

```python
from pathlib import Path

import chronopack as cp

REPO_ROOT = Path.cwd().resolve().parent if Path.cwd().name == "grn_inference" else Path.cwd().resolve()
ANALYSIS_DIR = REPO_ROOT / "grn_inference"
TABLE_DIR = ANALYSIS_DIR / "outputs" / "tables"
TF_DIR = ANALYSIS_DIR / "outputs" / "omnipath_tf_ranks"
OUTPUT_DIR = TABLE_DIR / "chronopack_grn"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

VALID_TARGETS_PATH = Path("/new-stg/home/alex/tmp/grn_inference_v2/reproduction/inputs/external/ref/genome/hg38/annotations/protein_coding_genes.csv")
GENE_SCORES_PATH = TABLE_DIR / "thp1_k562_lfdr_recalibration_scores.csv"
TF_SCORES = {
    "ChronoPack_K562": TF_DIR / "omnipath_ranks.TNFa.csv",
    "ChronoPack_THP1": TF_DIR / "omnipath_ranks.TLR4.csv",
}
CONTEXTS = {"ChronoPack_K562": "K562", "ChronoPack_THP1": "THP1"}
```

```python
grn = cp.tl.grn.infer(
    gene_scores=GENE_SCORES_PATH,
    tf_scores=TF_SCORES,
    valid_targets=VALID_TARGETS_PATH,
    contexts=CONTEXTS,
    provenance={"workflow": "grn_inference/ChronoPackRegulatoryEdges.ipynb"},
)
paths = grn.write(OUTPUT_DIR, prefix="chronopack_grn")
paths
```
