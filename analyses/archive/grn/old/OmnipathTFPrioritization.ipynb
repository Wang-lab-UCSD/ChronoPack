# OmniPath TF Prioritization

Ranks context-specific TFs using the package-backed ChronoPack OmniPath signed diffusion workflow.

```python
from pathlib import Path

import chronopack as cp

REPO_ROOT = Path.cwd().resolve().parent if Path.cwd().name == "grn_inference" else Path.cwd().resolve()
ANALYSIS_DIR = REPO_ROOT / "grn_inference"
OUTPUT_DIR = ANALYSIS_DIR / "outputs" / "omnipath_tf_ranks"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

HUMAN_TF_PATH = Path("/new-stg/home/alex/ref/gene_sets/hs_tfs.txt")
LIGAND_RECEPTOR_PATH = Path("/new-stg/home/alex/ref/cellchatdb/hs/interaction_input_CellChatDB.csv")

CONTEXTS = {
    "TNFa": {"source": "TNF", "is_ligand": True},
    "TLR4": {"source": "TLR4", "is_ligand": False},
}
```

```python
tf_result = cp.tl.grn.rank_regulators(
    human_tf_path=HUMAN_TF_PATH,
    ligand_receptor_path=LIGAND_RECEPTOR_PATH,
    contexts=CONTEXTS,
    output_dir=OUTPUT_DIR,
    w_signed=1.0,
    w_unsigned=0.5,
    alpha=0.85,
    max_steps=80,
    min_steps=10,
    tol_mass=1e-12,
)
tf_result.write(OUTPUT_DIR)
```

```python
for label, ranking in tf_result.rankings.items():
    print(label, ranking.shape)
    display(ranking.head(10))
```
