# K562 + TNF-alpha: two-lineage gene dynamics

Fits negative-binomial GAM gene dynamics for the two K562 trajectory clusters (lineage 1: inflammatory-high, lineage 2: erythroid-like) with ChronoPack's tradeSeq backend. Both lineages run along time; each cell enters them with its soft cluster memberships (`cell_weights.csv`, written by `03_PrepareForTradeSeq.ipynb`). A 4-knot screen of all candidate genes with global and per-lineage association tests is followed by an 8-knot refit of genes with p < 0.05 in any of them. Outputs (`asso_df.2trajectory.nknots4.csv`, `sce_screen.2trajectory.nknots4.rds`, `sce.2trajectory.nknots8.rds`, ...) are written next to the one-lineage fit in the input directory.

This fit is not used by `gene_clustering/`.

```python
import os
from pathlib import Path

import chronopack as cp

# Folder holding one sub-folder per experiment; see analyses/README.md.
DATA_DIR = Path(os.environ.get("CHRONOPACK_DATA_DIR", "../data"))
EXP_DIR = DATA_DIR / "k562-tnfa.hg38.rep"
```

```python
INPUT_DIR = (EXP_DIR / "intmd/sc-K562-1e4_to_tradeSeq.fixed").resolve()
INPUT_DIR
```

```python
fit = cp.tl.fit_gene_dynamics(
    input_dir=INPUT_DIR,
    backend="tradeseq",
    lineage_weights="cell_weights.csv",
    n_points=9,
    workers=4,
    screen_nknots=4,
    final_nknots=8,
    p_threshold=0.05,
)
```

```python
{
    "genes_tested": len(fit.association),
    "significant_q_lt_0.05_any_test": len(fit.significant),
    "refit_with_8_knots": len(fit.selected_genes),
    "outputs": fit.sidecar_paths,
}
```
