```python
import numpy as np
from scipy import sparse

def check_anndata_equivalence(a, b, rtol=1e-7, atol=1e-10, verbose=True):
    def matrix_equal(x, y):
        if x.shape != y.shape:
            return False

        if sparse.issparse(x) or sparse.issparse(y):
            x = sparse.csr_matrix(x)
            y = sparse.csr_matrix(y)
            diff = x - y

            if diff.nnz == 0:
                return True

            return np.all(
                np.isclose(diff.data, 0, rtol=rtol, atol=atol)
            )

        return np.allclose(
            np.asarray(x),
            np.asarray(y),
            rtol=rtol,
            atol=atol,
            equal_nan=True
        )

    results = {}

    # Dimensions / indices
    results["shape"] = a.shape == b.shape
    results["obs_index"] = a.obs_names.equals(b.obs_names)
    results["var_index"] = a.var_names.equals(b.var_names)

    # X
    results["X"] = matrix_equal(a.X, b.X)

    # Layers
    layers_a = set(a.layers.keys())
    layers_b = set(b.layers.keys())

    results["layer_keys"] = layers_a == layers_b

    for layer in sorted(layers_a | layers_b):
        if layer not in a.layers or layer not in b.layers:
            results[f"layer:{layer}"] = False
        else:
            results[f"layer:{layer}"] = matrix_equal(
                a.layers[layer],
                b.layers[layer]
            )

    equivalent = all(results.values())

    if verbose:
        for key, value in results.items():
            print(f"{key:20s}: {'✓' if value else '✗'}")
        print("-" * 30)
        print(f"Equivalent          : {'✓ YES' if equivalent else '✗ NO'}")

    return equivalent, results
```
