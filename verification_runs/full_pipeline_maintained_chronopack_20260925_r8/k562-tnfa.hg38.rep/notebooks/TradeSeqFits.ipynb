# K562 + TNF-alpha: gene dynamics

Fits negative-binomial GAM gene dynamics with ChronoPack's tradeSeq backend, using the inputs written by `PrepareForTradeSeq.ipynb`: a 4-knot screen with an association test over all candidate genes, then an 8-knot refit of genes with screen p < 0.05. Association tables and the fitted `SingleCellExperiment` objects (`sce_screen.1trajectory.nknots4.rds`, `sce.1trajectory.nknots8.rds`) are written to the input directory.


```python
from pathlib import Path

import chronopack as cp
```

```python
INPUT_DIR = Path("../intmd/sc-K562-1e4_to_tradeSeq.fixed").resolve()
INPUT_DIR
```

```python
fit = cp.tl.fit_gene_dynamics(
    input_dir=INPUT_DIR,
    backend="tradeseq",
    n_points=9,
    workers=4,
    screen_nknots=4,
    final_nknots=8,
    p_threshold=0.05,
    # Older file names that downstream notebooks still read.
    legacy_outputs={
        "association": ["asso_df.1trajectory.csv"],
        "significant": ["sig_df.1trajectory.csv"],
    },
)
```

```python
{
    "genes_tested": len(fit.association),
    "significant_q_lt_0.05": len(fit.significant),
    "refit_with_8_knots": len(fit.selected_genes),
    "outputs": fit.sidecar_paths,
}
```
